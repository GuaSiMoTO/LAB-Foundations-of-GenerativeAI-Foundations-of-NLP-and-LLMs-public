## Iteration 1: Loading and Configuring the Ollama Model
1. Ensure **Ollama** is running and accessible in your local environment.
2. Set up the necessary configurations for interacting with **Ollama**.
   - Define a function to send prompts to **Ollama** and receive the model's response.

```python
from ollama import Client

# 1. Ensure Ollama is running and accessible
OLLAMA_HOST = "http://localhost:11434"

# 2. Set up configurations and create client
ollama = Client(host=OLLAMA_HOST)

# Function to send prompts and get responses
def ask_ollama(prompt: str, model: str = "deepseek-r1:8b") -> str:
    try:
        response = ollama.generate(model=model, prompt=prompt)
        return response['response']
    except Exception as e:
        print(f"Error: {e}")
        return ""
```

## Iteration 2: Text Summarization with Ollama
1. Choose a text file (`article.txt`) containing substantial content for summarization.
2. Generate a summary using **Ollama**:
   - Design a prompt to instruct **Ollama** to summarize the text.
   - Define a function that sends the prompt and text to **Ollama** and retrieves the summary.
   - Save the summary to `summary_ollama.txt`.

```python
import re
# Load the text from the file
with open("article.txt", "r") as file:
    text_to_summarize = file.read()

# Define the prompt for summarization
prompt = f"""
Summarize the following text in a concise and clear manner, and format it with line breaks for readability. length max to 50 tokens:

{text_to_summarize}
"""

# Generate summary (USANDO tu función ask_ollama)
ollama_summary = ask_ollama(prompt)
def clean_think_tags(text):
    """
    Elimina todo el contenido entre <think> y </think> incluyendo las etiquetas
    """
    return re.sub(r'<think>.*?</think>', '', text, flags=re.DOTALL)

text_final = clean_think_tags(ollama_summary)

# Save the cleaned summary to a file
with open("summary_ollama.txt", "w") as file:
    file.write(text_final)
```
